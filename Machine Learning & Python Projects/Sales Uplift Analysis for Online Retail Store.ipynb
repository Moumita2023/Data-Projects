# 📊 Sales Uplift Analysis Tutorial

This notebook will guide you step by step through an **uplift analysis** to evaluate the effect of a marketing campaign.

We will:
1. Load and explore the dataset
2. Perform Exploratory Data Analysis (EDA)
3. Conduct an A/B statistical test (t-test)
4. Draw a business conclusion

---

```python
# 📌 Step 1: Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import ttest_ind
```

```python
# 📌 Step 2: Load the dataset
data_path = "sales_data_with_marketing_intervention.csv"  # Change this to your file name
df = pd.read_csv(data_path)

df.head()
```

### 🔍 Step 3: Understand the dataset
We want to check if the campaign had an impact on sales.
- `campaign = TRUE` → customers who received the intervention (treatment)
- `campaign = FALSE` → customers who did not (control)
- `revenue` → outcome metric to measure impact

👉 The key question: *Did customers in the campaign group spend more than those in the control group?*

```python
# Summary statistics grouped by campaign
df.groupby("campaign")["revenue"].describe()
```

## 📊 Step 4: Exploratory Data Analysis (EDA)
We will visualize the difference between **campaign** and **non-campaign** groups.

```python
# Average revenue per transaction
sns.barplot(x="campaign", y="revenue", data=df, estimator=np.mean, ci=None)
plt.title("Average Revenue per Transaction (Campaign vs Control)")
plt.show()

# Revenue distribution
sns.histplot(data=df, x="revenue", hue="campaign", kde=True, bins=30)
plt.title("Revenue Distribution (Campaign vs Control)")
plt.show()
```

👉 From these plots, you should start to see whether the campaign group spent more than the control group.

## 📈 Step 5: Statistical Test (A/B Testing)
EDA gives us a visual clue, but we need to know if the difference is **real or random chance**.

We’ll run a **t-test**:
- **Null hypothesis (H₀)**: The campaign has no effect (means are equal)
- **Alternative hypothesis (H₁)**: The campaign increased revenue

If `p < 0.05`, we reject H₀ and say the campaign had a significant effect.

```python
campaign_sales = df[df['campaign'] == True]['revenue']
control_sales = df[df['campaign'] == False]['revenue']

# T-test
t_stat, p_val = ttest_ind(campaign_sales, control_sales, equal_var=False)

# Calculate uplift percentage
uplift = (campaign_sales.mean() - control_sales.mean()) / control_sales.mean() * 100

print(f"Mean Revenue (Campaign): {campaign_sales.mean():.2f}")
print(f"Mean Revenue (Control): {control_sales.mean():.2f}")
print(f"Uplift: {uplift:.2f}%")
print(f"p-value: {p_val:.4f}")
```

## ✅ Step 6: Business Conclusion
- If `p < 0.05`: The campaign significantly increased revenue.
- If `p >= 0.05`: No evidence that the campaign had a real impact.

👉 Example wording for portfolio:

“The marketing campaign increased average customer revenue by **X%**. A t-test confirmed this uplift is statistically significant (p < 0.05), suggesting the campaign had a **real positive effect** on sales.”
