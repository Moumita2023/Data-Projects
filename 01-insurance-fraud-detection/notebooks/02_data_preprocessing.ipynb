## Insurance Claims Fraud Detection
### Notebook 2 — Data Preprocessing & Feature Engineering

#### Objective
Handle missing values, fix data types, and engineer new features
for fraud risk scoring based on domain knowledge of P&C insurance.

```python
import pandas as pd
import numpy as np

# Load the baseline clean files from Notebook 1
claims_df = pd.read_csv('../data/processed/claims_clean.csv')
employee_df = pd.read_csv('../data/processed/employee_clean.csv')
vendor_df = pd.read_csv('../data/processed/vendor_clean.csv')

print("Datasets loaded successfully")
print(f"Claims:   {claims_df.shape[0]:,} rows, {claims_df.shape[1]} columns")
print(f"Employee: {employee_df.shape[0]:,} rows, {employee_df.shape[1]} columns")
print(f"Vendor:   {vendor_df.shape[0]:,} rows, {vendor_df.shape[1]} columns")
```

```python
# Drop ADDRESS_LINE2 — over 83% missing across all datasets
# Not useful for fraud analysis

claims_df.drop(columns=['ADDRESS_LINE2'], inplace=True)
employee_df.drop(columns=['ADDRESS_LINE2'], inplace=True)
vendor_df.drop(columns=['ADDRESS_LINE2'], inplace=True)

print("✅ ADDRESS_LINE2 dropped from all three datasets")
```

```python
# Fill missing values in Claims Data

# CITY and INCIDENT_CITY — very few missing, fill with Unknown
claims_df['CITY'] = claims_df['CITY'].fillna('Unknown')
claims_df['INCIDENT_CITY'] = claims_df['INCIDENT_CITY'].fillna('Unknown')

# CUSTOMER_EDUCATION_LEVEL — fill with Unknown
claims_df['CUSTOMER_EDUCATION_LEVEL'] = claims_df['CUSTOMER_EDUCATION_LEVEL'].fillna('Unknown')

# AUTHORITY_CONTACTED — meaningful fraud signal
# No authority contacted = higher fraud risk
claims_df['AUTHORITY_CONTACTED'] = claims_df['AUTHORITY_CONTACTED'].fillna('None')

# VENDOR_ID — meaningful fraud signal
# No vendor assigned = claim not investigated
claims_df['VENDOR_ID'] = claims_df['VENDOR_ID'].fillna('No Vendor')

print("✅ Missing values handled in Claims Data")
print(f"Remaining missing values: {claims_df.isnull().sum().sum()}")
```

```python
# Fill missing values in Employee and Vendor Data
employee_df['CITY'] = employee_df['CITY'].fillna('Unknown')
vendor_df['CITY'] = vendor_df['CITY'].fillna('Unknown')

print("✅ Missing values handled in Employee and Vendor Data")
```

```python
# Final check — confirm all missing values handled
for name, df in [('CLAIMS', claims_df),
                 ('EMPLOYEE', employee_df),
                 ('VENDOR', vendor_df)]:
    remaining = df.isnull().sum().sum()
    status = "✅ No missing values" if remaining == 0 else f"⚠️ {remaining} missing values remain"
    print(f"{name}: {status}")
```

```python
# Save preprocessed files — these will be used for fraud scoring
claims_df.to_csv('../data/processed/claims_preprocessed.csv', index=False)
employee_df.to_csv('../data/processed/employee_preprocessed.csv', index=False)
vendor_df.to_csv('../data/processed/vendor_preprocessed.csv', index=False)

print("✅ Preprocessed datasets saved to data/processed/")
```
