```python
# Standard data manipulation libraries
import pandas as pd
import numpy as np

# Visualisation libraries
import matplotlib.pyplot as plt
import seaborn as sns

# Display settings
pd.set_option('display.max_columns', None)  # show all columns
pd.set_option('display.float_format', '{:.2f}'.format)  # clean decimals

print("Libraries loaded successfully")
```

```python
# Load all three datasets from raw data folder
claims_df = pd.read_csv('../data/raw/insurance_data.csv')
employee_df = pd.read_csv('../data/raw/employee_data.csv')
vendor_df = pd.read_csv('../data/raw/vendor_data.csv')

print("Datasets loaded successfully")
print(f"Claims Data:   {claims_df.shape[0]:,} rows, {claims_df.shape[1]} columns")
print(f"Employee Data: {employee_df.shape[0]:,} rows, {employee_df.shape[1]} columns")
print(f"Vendor Data:   {vendor_df.shape[0]:,} rows, {vendor_df.shape[1]} columns")
```

```python
# Cell 6 code
#  Check missing values in all three datasets
for name, df in [('CLAIMS', claims_df), 
                 ('EMPLOYEE', employee_df), 
                 ('VENDOR', vendor_df)]:
    
    missing = df.isnull().sum()
    missing_pct = (df.isnull().sum() / len(df) * 100).round(2)
    
    missing_df = pd.DataFrame({
        'Missing Count': missing,
        'Missing %': missing_pct
    })
    
    # Only show columns that actually have missing values
    missing_df = missing_df[missing_df['Missing Count'] > 0]
    
    print(f"\n{'=' * 60}")
    print(f"{name} DATA — Missing Values")
    print(f"{'=' * 60}")
    
    if missing_df.empty:
        print("✅ No missing values found")
    else:
        display(missing_df)
```

```python
# Cell 7 code
# Check for duplicate rows in each dataset
for name, df in [('CLAIMS', claims_df), 
                 ('EMPLOYEE', employee_df), 
                 ('VENDOR', vendor_df)]:
    
    dupes = df.duplicated().sum()
    status = "✅ No duplicates" if dupes == 0 else f"⚠️ {dupes} duplicates found"
    print(f"{name} DATA: {status}")
```

```python
# Cell 8 code
# Save cleaned datasets to processed folder
# We will add more cleaning steps in next cells
# but save a baseline version first

claims_df.to_csv('../data/processed/claims_clean.csv', index=False)
employee_df.to_csv('../data/processed/employee_clean.csv', index=False)
vendor_df.to_csv('../data/processed/vendor_clean.csv', index=False)

print("✅ Clean datasets saved to data/processed/")
```
